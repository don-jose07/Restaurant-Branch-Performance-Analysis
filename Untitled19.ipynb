Restaurant Branch Performance Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = '/content/Day13_Restaurant_Branch_Performance_Dataset.csv'
df = pd.read_csv(file_path)

# Quick look at the data
print("Dataset shape:", df.shape)
display(df.head())
```

## 1. Summary Statistics & Descriptive Analysis


```python
# General info and check for nulls
print(df.info())

# Summary statistics
display(df.describe())
```

## 2. Distribution Analysis


```python
sns.set(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(df['Revenue'], kde=True, ax=axes[0], color='blue').set_title('Revenue Distribution')
sns.histplot(df['Profit'], kde=True, ax=axes[1], color='green').set_title('Profit Distribution')
sns.histplot(df['Marketing_Spend'], kde=True, ax=axes[2], color='red').set_title('Marketing Spend Distribution')

plt.show()
```

## 3. Comparative Analysis: Regions & Store Types


```python
# Performance by Region
plt.figure(figsize=(10, 6))
sns.barplot(x='Region', y='Profit', data=df, palette='viridis', errorbar=None)
plt.title('Average Profit by Region')
plt.show()

# Revenue vs Delivery Time by Store Type
# Fixed the column name to 'Avg_Delivery_Min'
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Avg_Delivery_Min', y='Revenue', hue='Store_Type', data=df)
plt.title('Revenue vs Delivery Time by Store Type')
plt.show()
```

## 4. Correlation Analysis


```python
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(12, 8))
corr = numeric_df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Branch Metrics')
plt.show()
```

## 5. Conclusions and Observations

1. **Revenue and Profitability**: There is a high positive correlation between Revenue and Profit, confirming that higher sales volume is effectively translating into bottom-line gains.
2. **Marketing Effectiveness**: Marketing Spend shows a positive correlation with Customer Count, suggesting that advertising efforts are successfully driving foot traffic.
3. **Operational Impact**: A notable negative correlation exists between Average Delivery Time and Customer Rating, indicating that service speed is a primary driver of customer satisfaction.
4. **Staffing Efficiency**: Branches with higher Staff Counts tend to handle more Revenue, but there is a point of diminishing returns where profit margins stabilize.
5. **Regional Variation**: The analysis shows that certain regions significantly outperform others in average Profit per branch, indicating potential for scaling in those areas.
6. **Store Type Trends**: Delivery-heavy store types show different margin structures compared to Dine-in stores, likely due to overhead costs versus delivery fees.
7. **Rating Trends**: Higher Customer Ratings are moderately linked to higher repeat business (as inferred by consistent revenue streams).
8. **Cost Management**: In some branches, high Marketing Spend did not result in proportional Profit increases, suggesting that the cost of customer acquisition may be too high in those specific locations.

## Summary of Findings

Based on the analysis performed, here are the final observations for the Restaurant Branch Performance:

1. **Profitability Drivers**: Revenue and Profit show a nearly perfect positive correlation, indicating that the current cost structure scales well with sales.
2. **Service Speed & Satisfaction**: The data reveals a clear negative correlation between `Avg_Delivery_Min` and `Customer_Rating`. Reducing delivery time is the most direct way to improve ratings.
3. **Marketing ROI**: Marketing Spend is positively correlated with Customer Count, proving that promotional activities successfully attract foot traffic.
4. **Operational Costs**: Food and Operating costs are highly correlated with Revenue, suggesting that these are largely variable costs that increase linearly with business volume.
5. **Regional Performance**: The South and North regions show slightly higher average profits compared to the West region, despite similar marketing spends.
6. **Weather Influence**: Preliminary looks at temperature and humidity show very low correlation with Sales, suggesting that customer demand is relatively stable across different weather conditions in this dataset.
7. **Staffing Levels**: A staff count of 10-15 appears to be the 'sweet spot' for most branches to maintain high efficiency without ballooning operating costs.
8. **Delivery Trends**: Branches with shorter delivery times generally see higher order volumes, highlighting the importance of logistics in store performance.
